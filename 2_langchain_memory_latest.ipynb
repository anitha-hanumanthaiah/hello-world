```python
from dotenv import load_dotenv
import os

from langchain_google_genai import ChatGoogleGenerativeAI

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
# Load environment variables
load_dotenv()
```

```python
# Read API key
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")

# Create Gemini Model
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    google_api_key=GOOGLE_API_KEY,
    temperature=0.3
)
```

```python
# Prompt Template
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful AI assistant."),
        ("human", "{input}")
    ]
)
```

```python
# Output Parser
parser = StrOutputParser()

# LCEL Chain
chain = prompt | llm | parser

# Store Chat Sessions
store = {}
```

```python
# Function to get chat history
def get_session_history(session_id: str):

    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()

    return store[session_id]
```

```python
# Add Memory to Chain
chain_with_memory = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input"
)
```

```python
# Session ID
config = {
    "configurable": {
        "session_id": "user_1"
    }
}
```

```python
# Conversation 1
response_1 = chain_with_memory.invoke(
    {
        "input": "Hi, what's your name?"
    },
    config=config
)

print("\nAI:", response_1)
```

```python
# Conversation 2
response_2 = chain_with_memory.invoke(
    {
        "input": "My name is Sashi."
    },
    config=config
)

print("\nAI:", response_2)
```

```python
# Conversation 3
response_3 = chain_with_memory.invoke(
    {
        "input": "What's my name?"
    },
    config=config
)

print("\nAI:", response_3)
```

```python
# Conversation 4
response_4 = chain_with_memory.invoke(
    {
        "input": "My hobby is Reading."
    },
    config=config
)

print("\nAI:", response_4)
```
